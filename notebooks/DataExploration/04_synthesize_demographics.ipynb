# Cross-cohort demographics synthesis
Combines the completed ADNI, OASIS-3, and AIBL audits into cross-cohort comparison CSVs and one detailed Markdown report.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import Markdown, display
PROJECT_ROOT = Path.cwd().resolve()
while not (PROJECT_ROOT / 'datasets').exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
from scripts.demographics_pipeline import synthesize_report
print(f'Project root: {PROJECT_ROOT}')
print(f'Python: {sys.version.split()[0]}')
```

```python
report_path = synthesize_report(verbose=True)
print(f'Report written to: {report_path.relative_to(PROJECT_ROOT)}')
```

```python
combined_dir = PROJECT_ROOT / 'results' / 'csvs' / 'combined'
print('Cohort overview')
display(pd.read_csv(combined_dir / 'cohort_overview.csv'))
print('Cross-cohort baseline metrics')
display(pd.read_csv(combined_dir / 'cross_cohort_baseline_metrics.csv'))
print('Cross-cohort concept availability')
display(pd.read_csv(combined_dir / 'cross_cohort_concept_availability.csv'))
```

```python
report_text = report_path.read_text(encoding='utf-8')
print(f'Final report contains {len(report_text):,} characters.')
display(Markdown(report_text[:12000]))
print('Notebook preview is limited to 12,000 characters; open the Markdown report for its complete usable-column appendices.')
```
